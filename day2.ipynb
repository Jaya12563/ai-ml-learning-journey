```python
# Correlation matrix
import seaborn as sns

# Get numerical columns
numerical_df = df.select_dtypes(include=['float64', 'int64'])

# Calculate correlation
correlation = numerical_df.corr()

# Create heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0,
            cbar_kws={'label': 'Correlation'}, vmin=-1, vmax=1)

plt.title('Feature Correlation Matrix', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("Correlation Analysis:")
print("="*70)
print(correlation)
print("\nInterpretation:")
print("• Values close to 1: Features move together (positive correlation)")
print("• Values close to 0: Features are independent")
print("• Values close to -1: Features move opposite (negative correlation)")
print("\nKey insight:")
print("✓ PetalLength & PetalWidth are highly correlated (0.96)")
print("✓ Other features have moderate correlations")
print("✓ Good feature diversity for ML!")
print("="*70)
```
